```python
import re
import random

reflections = {
    "i": "you", "me": "you", "my": "your", "am": "are",
    "i'm": "you're", "you": "I", "your": "my", "you're": "I'm"
}

def reflect(fragment):
    words = fragment.lower().split()
    return " ".join(reflections.get(w, w) for w in words)

rules = [
    (r"i need (.*)", [
        "Why do you need {0}?",
        "Would it help you to have {0}?"
    ]),
    (r"i am (.*)", [
        "How long have you been {0}?",
        "Why do you think you are {0}?"
    ]),
    (r"i feel (.*)", [
        "Tell me more about feeling {0}.",
        "Why do you feel {0}?"
    ]),
    (r"i want (.*)", [
        "Why do you want {0}?",
        "What would it mean to you to have {0}?"
    ]),
    (r"(.*)\bmother\b(.*)|(.*)\bfather\b(.*)", [
        "Tell me more about your family.",
        "How do you feel about your family?"
    ]),
]

fallbacks = [
    "Tell me more about that.",
    "How does that make you feel?",
    "Why do you say that?",
]

def eliza_respond(text):
    text = text.lower().strip().rstrip('.')
    for pattern, responses in rules:
        match = re.search(pattern, text)
        if match:
            response = random.choice(responses)
            if "{0}" in response:
                captured = match.group(1) if match.groups() and match.group(1) else ""
                return response.format(reflect(captured))
            else:
                return response
    return random.choice(fallbacks)
```

```python
def chat():
    print("ELIZA: Hello. What is on your mind today? Type 'bye' to quit.")
    while True:
        user_input = input("You: ")
        if user_input.lower().strip() in ("bye", "quit", "exit"):
            print("ELIZA: Goodbye. Take care.")
            break
        print("ELIZA:", eliza_respond(user_input))

chat()
```
